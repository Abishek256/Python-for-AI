```python
#Implementation of Stack
stack = []
top = -1

# PUSH operation
def push(data):
    global top
    stack.append(data)
    top += 1
    print(data, "inserted")

# POP operation
def pop():
    global top
    if top == -1:
        print("Stack is Empty")
    else:
        print("Deleted element:", stack[top])
        stack.pop()
        top -= 1

# PEEK operation
def peek():
    if top == -1:
        print("Stack is Empty")
    else:
        print("Top element:", stack[top])

# Function calls
push(10)
push(20)
push(30)
print("Stack:", stack)
peek()
pop()
print("Stack after pop:", stack)
peek()
```

```python
#Implementation of LinkedList
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
head = None
tail = None
# First node
newNode = Node(10)
if head == None and tail == None:
    head = newNode
    tail = newNode
# Second node
newNode = Node(20)
tail.next = newNode
tail = newNode
# Third node
newNode = Node(30)
tail.next = newNode
tail = newNode
# Fourth node
newNode = Node(40)
tail.next = newNode
tail = newNode
# Display linked list
temp = head
while temp != None:
    print(temp.data, end=" -> ")
    temp = temp.next
print("None")
```
