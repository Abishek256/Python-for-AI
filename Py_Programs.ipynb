```python
def count_frequency(arr):
    freq = {}
    for num in arr:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1
    for num, count in freq.items():
        print(num, ":", count)
arr = [1, 2, 3, 2, 1, 4, 2, 3]
count_frequency(arr)
```

```python
def find_repeating(arr):
    freq = {}
    for num in arr:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1
    for i in freq:
      if freq[i] > 1:
        print(i)
arr = [1, 2, 3, 2, 1, 4, 2, 3]
find_repeating(arr)

```

```python
def find_non_repeating(arr):
    freq = {}
    for num in arr:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1
    for i in freq:
        if freq[i] == 1:
            print(i)
arr = [1, 2, 3, 2, 1, 4, 2, 3,5]
find_non_repeating(arr)
```
