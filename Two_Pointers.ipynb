```python
#Reverse an array
def reverse_array(arr):
  left = 0
  right = len(arr) - 1
  while left < right:
    arr[left],arr[right] = arr[right],arr[left]
    left += 1
    right -= 1
  return arr
reverse_array([1,2,3,4,5])
```

```python
#Find Missing Number
def missing_number(arr):
  n = len(arr)
  range_sum = n*(n + 1) // 2
  arr_sum = sum(arr)
  missing = range_sum - arr_sum
  return missing
missing_number([0,1,2,4,5])
```
