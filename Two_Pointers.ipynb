```python
#Reverse an array
def reverse_array(arr):
  left = 0
  right = len(arr) - 1
  while left < right:
    arr[left],arr[right] = arr[right],arr[left]
    left += 1
    right -= 1
  return arr
reverse_array([1,2,3,4,5])
```

```python
#Find Missing Number
def missing_number(arr):
  n = len(arr)
  range_sum = n*(n + 1) // 2
  arr_sum = sum(arr)
  missing = range_sum - arr_sum
  return missing
missing_number([0,1,2,4,5])
```

```python
#Remove Element
def remove_element(nums,val):
  j=0
  for i in range(len(nums)):
    if nums[i] != val:
      nums[j] = nums[i]
      j += 1
  return j

remove_element([1,2,2,3,4,5,6],2)
```

```python
#Remove Duplicates from Sorted Array
def remove_duplicate(nums):
  j=1
  for i in range(1,len(nums)):
    if nums[i] != nums[i - 1]:
      nums[j] = nums[i]
      j += 1
  return j
remove_element([6,6,8,8,9,9,9,12])
```

```python
#Move Zeros
def move_Zeros(nums):
  j = 0
  for i in range(len(nums)):
    if nums[i] != 0:
      nums[j] = nums[i]
      j += 1
  for i in range(j,len(nums)):
    nums[i] = 0
  return nums
move_Zeros([1,0,5,2,0,8,0,9])
```
